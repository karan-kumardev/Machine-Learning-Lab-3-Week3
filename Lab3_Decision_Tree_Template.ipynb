scikit-learn models require numeric input. Separate your target from your features, then convert any remaining categorical columns into numeric form.

**Tasks**

```python
# Task 3 — define y and X

# columns_to_remove = ["Churn Value", "Churn Label", "Churn Score", "CLTV", "Churn Reason"]
# df = df.drop(columns=columns_to_remove, errors='ignore')

#df.to_excel('clean_chunk.xlsx', index=False)


df_original=pd.read_excel('Telco_customer_churn.xlsx')
y=df_original['Churn Value']

X=pd.read_excel('clean_chunk.xlsx')

print("X(Feature) =",X.shape)
print("Y(Shape) =",y.shape)
```

```python
#Task 4

X['Gender'] = X['Gender'].map({'Male': 1, 'Female': 0})
yes_no_cols = ['Partner', 'Dependents', 'Phone Service', 'Paperless Billing']
for col in yes_no_cols:
    if col in X.columns:
        X[col]=X[col].map({'Yes': 1, 'No': 0})

multi_class_cols=['Multiple Lines', 'Internet Service', 'Online Security', 
                    'Online Backup', 'Device Protection', 'Tech Support', 
                    'Streaming TV', 'Streaming Movies', 'Contract', 'Payment Method']
X = pd.get_dummies(X, columns=[c for c in multi_class_cols if col in X.columns], drop_first=True, dtype=int)

if 'Senior Citizen' in X.columns:
    X['Senior Citizen'] = X['Senior Citizen'].astype(str).str.strip().map({'Yes': 1, 'No': 0, '1': 1, '0': 0}).astype(int)
    
```

```python
# Task 5 — confirm X is fully numeric with no missing values
print(X.dtypes)
```

**Answer 4 (encoding method & justification):**
the use of encoding method is changed for different columns because for those columns which contained only two categories, we used the manual mapping, while for those which had catogries more than 2, so we used one hot mapping.
